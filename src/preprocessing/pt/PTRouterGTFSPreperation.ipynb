> ### 💡 Info: Data Preparation for FleetPy Raptor Router
> 
> This notebook guides you through cleaning and formatting raw **GTFS data** into the specific file formats required by **FleetPy's Raptor Router**.
>
> #### 📄 Input Specifications:
> * **Mandatory GTFS Files:** `agency`, `calendar`, `routes`, `stop_times`, `stops`, `transfers`, and `trips`.
> * **Optional GTFS Files:** `calendar_dates`.
> * **Custom Required Files:** 
>   * `stations` (General requirement)
>   * `street_station_transfers` (Required for the **PTBrokerTPCS** module)
>
> **Note:** Any files not mentioned above will be ignored by the FleetPy reader.
>
> #### 📝 Output Naming Convention:
> To distinguish processed files from the originals, the suffix `_fp` will be appended to each filename:  
> `{original_filename}.txt` $\rightarrow$ `{original_filename}_fp.txt`
>
> #### 📂 Output Directory Structure:
> All processed files must be placed in the following directory path:
>
> `data/pt/{network_name}/{gtfs_name}/matched`
>
> * **`{network_name}`**: The name of the road network used in your FleetPy experiment.
> * **`{gtfs_name}`**: A custom identifier (string) that must be defined in your scenario configuration.

# Custom Required Files

### stations

```python
# read example stations file
stations_fp = pd.read_csv(os.path.join(example_gtfs_path, 'stations_fp.txt'))
print("\nStations file preview:")
print(stations_fp.head())
```

> #### 🚉 Stations File Requirements (Custom Input)
> 
> **Role in FleetPy:**
> Stations serve as the **interface points** between the Road Network and the PT Network.
> * **Passenger Flow:** Street Network Node $\rightarrow$ Walk $\rightarrow$ Station $\rightarrow$ Transfer $\rightarrow$ Specific PT Stops (within the station).
> 
> **Mandatory Columns:**
> * `station_id`
> * `stops_included`
> * `station_stop_transfer_times`
> * `num_stops_included`
> 
> **Optional Columns:**
> * `station_name`
> * `station_lat`, `station_lon`
> 
> **Detailed Column Specifications:**
> * **`station_id`**: A unique `string` used to index the station.
> * **`stops_included`**: A **list** containing all stop IDs belonging to this station. These IDs must exist in `stops_fp.txt`.
> * **`station_stop_transfer_times`**: A **list** representing the walking time (in **seconds**) from the abstract Station Node to each corresponding Stop Node.
> * **`num_stops_included`**: An `integer` representing the total number of stops in the station (i.e., the length of `stops_included`).
> * **`station_lat` / `station_lon`**: Coordinates of the station. It is recommended that the **CRS** (Coordinate Reference System) matches the network CRS used in your FleetPy experiment.
> 
> ⚠️ **Implementation Note:**
> * **Resolution:** Even if your GTFS data (stops_fp.txt) is already at the "station level" (rather than stop level), this file **must still be created**. In this case, `station_stop_transfer_times` can be a list containing only zeros.
> * **Backend Usage:** This file is explicitly called by the `_get_included_stops_and_transfer_times` method in **`RaptorRouterCpp`**.

### street_station_transfers

```python
# read example street_station_transfers file
street_station_transfers_fp = pd.read_csv(os.path.join(example_gtfs_path, 'street_station_transfers_fp.txt'))
print("\nStreet_station_transfers file preview:")
print(street_station_transfers_fp.head())
```

> #### 🛣️ Street-Station Transfers File Requirements
> *Target Module: PTBrokerTPCS*
> 
> **Mandatory Columns:**
> * `node_id`
> * `closest_station_id`
> * `street_station_transfer_time`
> 
> **Usage Context:**
> This file is **mandatory** for the `PTBrokerTPCS` module when the `_find_transfer_info` method is set to the `"closest"` query strategy (which is the default setting).
> 
> **Data Logic:**
> * **`node_id`**: Represents the PUDO (Pick-up/Drop-off) nodes.
> * **`closest_station_id`**: Records the ID of the nearest station node relative to the PUDO node.
> * **`street_station_transfer_time`**: Represents the walking time from the PUDO node to the closest station **in seconds**.
> 
> ℹ️ **Optimization Note:**
> If you are exclusively using `RaptorRouterCpp` to run the PT Router (without the PTBrokerTPCS module), this file is unnecessary and can be **ignored**.
