```python
scenario_name = 'test_manhattan_scenario_1'
base_dir = f'data/{scenario_name}/processed'
```

```python
dfs = {}
for f in os.listdir(base_dir):
    if f.endswith('.parquet'):
        file_path = os.path.join(base_dir, f)
        dfs[os.path.splitext(f)[0]] = pd.read_parquet(file_path)
        print(f"Loaded {f} into dfs['{os.path.splitext(f)[0]}']")
```
